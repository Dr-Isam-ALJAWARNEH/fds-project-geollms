```python
import os
import pandas as pd
import folium
from folium.plugins import HeatMap
from sklearn.preprocessing import MinMaxScaler
import shutil

# Load the recent NDVI file from GitHub
ndvi_url = "https://raw.githubusercontent.com/Dr-Isam-ALJAWARNEH/fds-project-geollms/main/Datasets/NDVI/NDVI20230301.csv"
ndvi_df = pd.read_csv(ndvi_url)

# Drop missing values
ndvi_df.dropna(subset=["x", "y", "grid_code"], inplace=True)

# Normalize NDVI (grid_code column)
scaler = MinMaxScaler()
ndvi_df["NDVI_scaled"] = scaler.fit_transform(ndvi_df[["grid_code"]])

# Prepare heatmap data (latitude, longitude, weight)
heat_data = ndvi_df[["y", "x", "NDVI_scaled"]].values.tolist()

# Create folium map centered on Chicago
ndvi_map = folium.Map(location=[41.8781, -87.6298], zoom_start=11)

# Add heatmap layer
HeatMap(heat_data, radius=8, max_zoom=13).add_to(ndvi_map)

# Save map to HTML
ndvi_map.save("ndvi_greenest_map_20230301.html")

# Save the file to Google Drive or download it directly
from google.colab import files
files.download("ndvi_greenest_map_20230301.html")
```
