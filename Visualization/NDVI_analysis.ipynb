```python
import pandas as pd
import folium
from folium import CircleMarker
from sklearn.preprocessing import MinMaxScaler

# Step 1: Load recent NDVI dataset
file_path = "fds-project-geollms/Datasets/NDVI/NDVI20230301.csv"
ndvi_df = pd.read_csv(file_path, usecols=['x', 'y', 'grid_code'])  # x = Longitude, y = Latitude, grid_code = NDVI

# Step 2: Drop missing values
ndvi_df.dropna(subset=['x', 'y', 'grid_code'], inplace=True)

# Step 3: Normalize NDVI values to 0-1 for coloring
scaler = MinMaxScaler()
ndvi_df['ndvi_scaled'] = scaler.fit_transform(ndvi_df[['grid_code']])

# Step 4: Create Folium map centered on Chicago
ndvi_map = folium.Map(location=[41.8781, -87.6298], zoom_start=11)

# Step 5: Add each point as a green circle, scaled by NDVI value
for _, row in ndvi_df.iterrows():
    ndvi_value = row['ndvi_scaled']
    color_intensity = int(255 * ndvi_value)
    color_hex = f'#00{color_intensity:02x}00'  # RGB for green gradient

    CircleMarker(
        location=[row['y'], row['x']],
        radius=3,
        color=color_hex,
        fill=True,
        fill_color=color_hex,
        fill_opacity=0.7
    ).add_to(ndvi_map)

# Step 6: Save and show
ndvi_map.save("ndvi_circle_visualization.html")
ndvi_map
```
