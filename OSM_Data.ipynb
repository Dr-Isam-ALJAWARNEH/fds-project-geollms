```python
import osmnx as ox
import pandas as pd

# Define the location
place_name = "Chicago, Illinois, USA"

# Walking Network
print("Downloading walking network")
G_walk = ox.graph_from_place(place_name, network_type='walk')
nodes_walk, edges_walk = ox.graph_to_gdfs(G_walk)

# Save edges with all attributes
edges_walk.to_csv("chicago_walk_edges_all_columns.csv", index=False)

# Biking Network
print("Downloading biking network")
G_bike = ox.graph_from_place(place_name, network_type='bike')
nodes_bike, edges_bike = ox.graph_to_gdfs(G_bike)

# Save edges with all attributes
edges_bike.to_csv("chicago_bike_edges_all_columns.csv", index=False)

print("CSV files saved successfully.")
```
