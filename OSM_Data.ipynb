```python
import osmnx as ox
import pandas as pd

# Define the location
place_name = "Chicago, Illinois, USA"

# Walking Network
print("Downloading walking network")
G_walk = ox.graph_from_place(place_name, network_type='walk')
nodes_walk, edges_walk = ox.graph_to_gdfs(G_walk)

# Save edges with all attributes
edges_walk.to_csv("chicago_walk_edges_all_columns.csv", index=False)

# Biking Network
print("Downloading biking network")
G_bike = ox.graph_from_place(place_name, network_type='bike')
nodes_bike, edges_bike = ox.graph_to_gdfs(G_bike)

# Save edges with all attributes
edges_bike.to_csv("chicago_bike_edges_all_columns.csv", index=False)

print("CSV files saved successfully.")
```

```python
import pandas as pd

# Load Full Walk CSV
walk_df = pd.read_csv("chicago_walk_edges_all_columns.csv")

# Load Full Bike CSV
bike_df = pd.read_csv("chicago_bike_edges_all_columns.csv")

# Define Useful Columns
useful_columns = [
    'u',            # start node ID
    'v',            # end node ID
    'key',          # edge key
    'name',         # street or path name
    'highway',      # type of path (e.g., footway, residential, cycleway)
    'length',       # length in meters
    'geometry'      # line geometry (shapely LineString)
]

# Filter, drop fully empty columns, and save Walk Network
walk_filtered = walk_df[walk_df.columns.intersection(useful_columns)]
walk_filtered = walk_filtered.dropna(axis=1, how='all')
walk_filtered.to_csv("chicago_walk_edges_useful_2.csv", index=False)

# Filter, drop fully empty columns, and save Bike Network
bike_filtered = bike_df[bike_df.columns.intersection(useful_columns)]
bike_filtered = bike_filtered.dropna(axis=1, how='all')
bike_filtered.to_csv("chicago_bike_edges_useful_2.csv", index=False)

print("Filtered files saved:")
print("chicago_walk_edges_useful.csv")
print("chicago_bike_edges_useful.csv")
```
