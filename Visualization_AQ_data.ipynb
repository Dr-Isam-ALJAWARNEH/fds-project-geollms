```python
import pandas as pd
import requests
from io import StringIO
import folium
from folium.plugins import HeatMap

# Step 1: Download and combine datasets
base_url = "https://raw.githubusercontent.com/Dr-Isam-ALJAWARNEH/fds-project-geollms/main/Datasets/AQ_data/"
filenames = [f"chicago_eclipse_data_part_{i}.csv" for i in range(1, 20)]
dfs = []

for filename in filenames:
    url = base_url + filename
    response = requests.get(url)
    if response.status_code == 200:
        df = pd.read_csv(StringIO(response.text))
        dfs.append(df)
    else:
        print(f"Failed to load: {filename}")

combined_df = pd.concat(dfs, ignore_index=True)

# Step 2: Drop rows with missing values in key columns
combined_df.dropna(subset=['Latitude', 'Longitude', 'PM25'], inplace=True)

# Step 3: Prepare data for heatmap (weight = PM2.5 level)
heat_data = combined_df[['Latitude', 'Longitude', 'PM25']].values.tolist()

# Step 4: Create folium map centered around Chicago
chicago_map = folium.Map(location=[41.8781, -87.6298], zoom_start=11)

# Step 5: Add heatmap layer
HeatMap(heat_data, radius=8, max_zoom=13).add_to(chicago_map)

# Step 6: Save or display
chicago_map.save("pm25_heatmap_chicago.html")
chicago_map
```
